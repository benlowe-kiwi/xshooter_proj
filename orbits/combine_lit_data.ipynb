# Extracting Gaia DR3 source IDs for stars

### Norris et al. (2013)
This has numerous naming schemes that need to be made aware of before putting into SIMBAD:

- **52972-5213-5047** This is an internal SDSS observation ID, structured as "MJD - Plug plate - fibre". Use that to get the RA and DEC coordinates for star, then to get UCAC4 ID, which can be queried on SIMBAD.
- **BS16545-089** This is part of the HK Survey. To Query, need to add "BPS" to the front, alongside spacing out "BS". As an example, this would be "BPS BS 16545-089".
- **CS30336-049** Again, this is part of the HK Survey. Like previously, need to add "BPS" to the front, alongside spacing out "CS". As an example, this would be "BPS CS 30336-049"
- **HE0049-3948** This is part of the HES Survey. To query, space the letters out from the numbers. As an example, this would be "HE 0049-3948"

Separating out the different stars in Norris et al. (2013)

```python
def get_sdss(star_id):
    """ Extracting UCAC4 ID from SDSS using plate-mjd-fibre format star ID.
    
    Parameters
    ----------
    star_id : str
        Star ID in plate-mjd-fibre format (e.g., '1234-56789-0123')
    
    Returns
    -------
    simbad_id : str
        Corresponding Simbad MAIN_ID for the star.
    """    
    # First getting the plate number, MJD and fibre number
    mjd = star_id.split('-')[0]
    plate_num = star_id.split('-')[1]
    fibre_num = star_id.split('-')[2]
    
    # Storing into Table
    sdss_target = Table({
    'plate': [plate_num], 
    'mjd': [mjd], 
    'fiberID': [fibre_num]
    })


    # We need to build a query to get RA/Dec for these objects.
    # The most efficient way for a list is to use an SQL query via astroquery.

    # 1. format the WHERE clause for multiple objects
    where_clauses = []
    for row in sdss_target:
        where_clauses.append(
            f"(plate={row['plate']} AND mjd={row['mjd']} AND fiberID={row['fiberID']})"
        )
    where_string = " OR ".join(where_clauses)

    # 2. Run the SQL query on SDSS Data Release 17 (or latest)
    query = f"""
    SELECT plate, mjd, fiberID, ra, dec
    FROM specObj
    WHERE {where_string}
    """

    sdss_results = SDSS.query_sql(query)
    
    # Extracting the RA and Dec values
    ra = sdss_results['ra']
    dec = sdss_results['dec']
    
    # Getting Simbad ID so to query
    c = SkyCoord(ra=ra*u.deg, dec=dec*u.deg)
    simbad_id = Simbad.query_region(c, radius=2 * u.arcsec)['MAIN_ID'][0]
    
    return simbad_id
    

# Running through Yong2013 data to get Gaia DR3 IDs
yong2013_dicts = []
for i, row in yong2013_rvs.iterrows():
    
    star = row['Name']
    rv = row['HRV']
    
    print(f"Running {i+1} of {len(yong2013_rvs)}")
    
    
    # --STEP 1: TIDY NOMENCLETURE--- $
    # Tidying SDSS stars
    # Need to first identify which stars are SDSS stars, then extract their Simbad IDs
    # by querying SDSS for their RA/Dec and then querying Simbad for their MAIN_IDs
    if star[0] == '5':
        
        # First extracting Simbad-compatible ID
        # For stars with no SDSS data, skip
        try:
            new_id = get_sdss(star)
            print(new_id)
        except TypeError:
            print(f"No SDSS data found for {star}")
            continue
            
            
    # Tidying up BPS BS stars
    elif star.startswith('BS'):
        new_id = f"BPS BS {star.split('BS')[1]}"
        star = f"BS {star.split('BS')[1]}"
            
    
    # Tidying up BPS CS stars
    elif star.startswith('CS'):
        new_id = f"BPS CS {star.split('CS')[1]}"
        star = f"CS {star.split('CS')[1]}"
        
        
    # Tidying up HE stars
    elif star.startswith('HE'):
        new_id = f"HE {star.split('HE')[1]}"
        star = f"HE {star.split('HE')[1]}"
    
    
    # ---STEP 2: QUERY SIMBAD TO GET GAIA DR3 ID--- $
    Simbad.add_votable_fields('ids')
    
    # Extracting all IDs associated with star on Simbad
    # If no data found, skip
    try:
        ids = Simbad.query_object(new_id)[0]
    except TypeError:
        print(f"No Simbad data found for {new_id}")
        continue
    
    # Extracting Gaia DR3 ID from list of IDs
    gaia_id = [s for s in str(ids['IDS']).split('|') if 'Gaia DR3' in s]
     
    # ---STEP 3: STORING RESULTS--- $
    if len(gaia_id) > 0:
        star_dict = {
            'catalog': 'Yong2013',
            'star_id': star,
            'gaia_id': gaia_id[0],
            'rv': rv
        }
        yong2013_dicts.append(star_dict)
    else:
        print(f"No Gaia DR3 ID found for {new_id}")
        continue
    
# Converting to DataFrame
yong2013_gaia_rvs = pd.DataFrame(yong2013_dicts)
display(yong2013_gaia_rvs)
```
