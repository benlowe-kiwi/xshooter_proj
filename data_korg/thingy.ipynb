```python
import numpy as np
import pandas as pd
```

```python
# Getting the P113 and P115 stellar parameters
p113_params = pd.read_csv("/data/misfit/benlowe/xshooter/p113/korg_data/abunds/final_iter9/stellar_params.csv")
p113_params["p#"] = "p113"
p115_params = pd.read_csv("/data/misfit/benlowe/xshooter/p115/korg_data/abunds_v2/final_iter9/stellar_params.csv")
p115_params["p#"] = "p115"

# Combining the two dataframes
combined_params = pd.concat([p113_params, p115_params], ignore_index=True)
combined_params.to_csv("/data/misfit/benlowe/xshooter/p115/korg_data/all_abunds/stellar_params_combined.csv", index=False)
```
