```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
from astropy.io import fits
from astropy.table import Table
```

Loading in the data

```python
# X-Shooter data
p113_data = pd.read_csv('/data/misfit/benlowe/xshooter/p113/korg_data/abunds/final_iter9/stellar_params.csv')
p115_data = pd.read_csv('/data/misfit/benlowe/xshooter/p115/korg_data/abunds_v2/final_iter9/stellar_params.csv')
comb_xshooter_data = pd.concat([p113_data, p115_data], ignore_index=True)

# Comparision literature data
lit_data = pd.read_csv('/data/misfit/benlowe/xshooter/lit_abund_data/all_lit_combined_data_v2.csv')
```

```python
len(comb_xshooter_data[['star_id', 'teff', 'logg', 'Fe_it1']])
```

Making the Kiel Diagram

```python
plt.figure(figsize=(8, 6))
plt.scatter(lit_data['teff'], lit_data['logg'], marker='.', color='grey', linewidths=0)
plt.scatter(comb_xshooter_data['teff'], comb_xshooter_data['logg'], marker='*', edgecolors='black', s=150, c=comb_xshooter_data['Fe_it1'], cmap='magma')

plt.colorbar(label='[Fe/H]')
plt.xlabel('Teff')
plt.ylabel('log g')
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
```
