```python
def html_plotter(x_data, y_data, x_label, y_label, title, filename, path):
    """Creates and saves interactive .html plots using the bokeh package.

    Parameters
    ----------
    x_data : ndarray
        Data on the x-axis
    y_data : ndarray
        Data on the y-axis
    x_label : str
        Label for the x-axis
    y_label : str
        Label for the y-axis
    title : str
        Title to be put at the top of the plot
    filename : str
        Name of the saved .html file.
    path : str
        Location for where to save the plot

    Returns
    -------
    None
    """
    # Defining the bokeh figure
    p = bok.figure(title=title, x_axis_label=x_label, y_axis_label=y_label, width = 1000)
    p.sizing_mode = 'scale_height'
    p.xaxis.axis_label_text_font_size = "12pt"
    p.yaxis.axis_label_text_font_size = "12pt"
    p.xaxis.major_label_text_font_size = "12pt"
    p.yaxis.major_label_text_font_size = "12pt"
    p.title.text_font_size = "16pt"

    # Plotting the spectra
    p.line(x_data, y_data, line_width=1.3, color='black')

    # Saving the spectra to the given directory
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", UserWarning)
        bok.save(obj=p, filename=path + filename, title=title)
    
    return None



# Loading in the normalised spectra files
path = '/data/misfit/benlowe/xshooter/p115/suppnet_data/normalised/'
savepath = '/data/misfit/benlowe/xshooter/p115/suppnet_data/final/'
files = [f for f in os.listdir(path) if f.endswith('.all')]

# Looping through each file, reading the data, plot, then save data to .csv
for file in files:
    
    # Loading in the file (with space delimiter)
    ob_df = pd.read_csv(path+file, delimiter=' ')
    
    # Extracting the star_id, arm and observation number from the filename
    star_id = file.split('_')[0] + '_' + file.split('_')[1] + '_' + file.split('_')[2] 
    arm = file.split('_')[3]
    ob_num = file.split('_')[4].split('.')[0]
    
    print(f'Processing {star_id} {arm} {ob_num}...')
    
    # Creating save directories
    ob_savepath = savepath+star_id+'/'+arm+'/'
    os.makedirs(ob_savepath, exist_ok=True)
    
    # Saving the dataframe to .csv
    ob_df.to_csv(ob_savepath+file+'.csv', index=False)
    
    # Making interactive plot to visualise the normalised spectra
    html_plotter(x_data=ob_df['wave'],
                 y_data=ob_df['normed_flux'],
                 x_label='Wavelength (Å)',
                 y_label='Normalised Flux',
                 title=f'{star_id} ({arm.upper()} {ob_num})',
                 filename=f'{file}.html',
                 path=ob_savepath)
    
```
