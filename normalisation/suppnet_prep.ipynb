```python
import numpy as np
import pandas as pd
from astropy.table import Table
import os
# from suppnet import get_suppnet
```

Loading in the list of files wanting to have normalised

```python
def extract_then_save(files_list, save_path, star_id, arm):
    
    for i, file in enumerate(files_list):
        ob = i + 1
        
        # Loading the data
        data = pd.read_csv(file)
        
        wave = data['wave'].values
        flux = data['flux'].values
        flux_err = data['flux_err'].values
        
        # Saving to a .dat file, with entries having a space separator
        entry = np.column_stack((wave, flux, flux_err))
        save_name = f"{star_id}_{arm}_ob{ob}.dat"
        np.savetxt(save_path + save_name, entry, fmt='%g', delimiter=' ')
        
        
        
        # save_name = f"{star_id}_{arm}_ob{ob}.csv"
        # data.to_csv(save_path + save_name, index=False)
    
    return None
        

# Getting the list of stars to be normalised
path = '/data/misfit/benlowe/xshooter/p115/reduced_data/p115_data/science_spec/'
stars = [star for star in os.listdir(path)]
save_path = '/data/misfit/benlowe/xshooter/p115/suppnet_data/raw/'

# For each star, open up the datasets (UVB, VIS, NIR)
for star in stars:
    star_path = path + star + '/'
    
    # For UVB
    uvb_path = star_path + 'uvb/'
    uvb_files = [uvb_path + f for f in os.listdir(uvb_path) if f.endswith('.csv')]
    extract_then_save(files_list=uvb_files, save_path=save_path, star_id=star, arm='uvb')
    
    # For VIS
    vis_path = star_path + 'vis/'
    vis_files = [vis_path + f for f in os.listdir(vis_path) if f.endswith('.csv')]
    extract_then_save(files_list=vis_files, save_path=save_path, star_id=star, arm='vis')
    
    # For NIR
    nir_path = star_path + 'nir/'
    nir_files = [nir_path + f for f in os.listdir(nir_path) if f.endswith('.csv')]
    extract_then_save(files_list=nir_files, save_path=save_path, star_id=star, arm='nir')
    
    
    
```
